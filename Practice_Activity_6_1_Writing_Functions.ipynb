0\. Complete the [**check-in** from Section 6.3.2](https://ds-ml-with-python.github.io/Course-Textbook/05-function_writing.html#nested-environments-and-scope): Write a custom function that does the following:

Limit the penguins dataset to a user-chosen species and/or island.
Makes a scatterplot of the penguin bill length and depth.
Try writing a version of this function using dynamic lookup, and a version where everything the function needs is passed in directly

```python
penguins = load_penguins()
def plot_bills_dynamic(species = None, island = None):
    penguin_subset = penguins
    
    if species is not None:
        penguin_subset = penguin_subset[penguin_subset["species"] == species]
    if island is not None:
        penguins_subset = penguin_subset[penguin_subset["island"] == island]
    penguin_subset = penguins_subset.dropna(subset = ['bill_length_mm', 'bill_depth_mm'])
    #specify the plot
    plot = (ggplot(penguins_subset, aes(x = 'bill_length_mm', y = 'bill_depth_mm')) + 
    geom_point() +
    labs(x= "Bill Length (mm)", y = "Bill Depth (mm)", title = f"Species is = {species} Island is = {island}"))
    return plot
```

```python
plot_bills_dynamic(species="Gentoo", island = "Biscoe")
```

```python
def plot_bills(dataframe, species = None, island = None):

    penguin_subset = dataframe
    if species is not None:
        penguin_subset = penguin_subset[penguin_subset["species"] == species]
    if island is not None:
        penguin_subset = penguin_subset[penguin_subset["island"] == island]
    penguin_subset = penguin_subset.dropna(subset = ['bill_length_mm', 'bill_depth_mm'])
    #specify the plot
    plot = (ggplot(penguin_subset, aes(x = 'bill_length_mm', y = 'bill_depth_mm')) + 
    geom_point() +
    labs(x= "Bill Length (mm)", y = "Bill Depth (mm)", title = f"Species is = {species} Island is = {island}"))
    return plot
```

```python
penguins = load_penguins()
plot_bills(penguins, species = "Adelie")
```
